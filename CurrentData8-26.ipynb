## August 2026 At-211 Run Current Data

```python
%matplotlib ipympl 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.dates as mdates

path = 'C:/users/willm/At-211/Current Data/Current_Aug27morning_2026.dat'
df = pd.read_csv(path, skiprows = 1, header = None, names = [ 'date', 'time', 'na1','na2','na3', 'current', 'dashes', 'node'], sep = ' ')
data = df[['date','time','current']]

# data['timestamp'] = pd.to_datetime(data['date'] + ' ' + data['time'], format='%m/%d/%Y %H:%M:%S')
# data['timestamp'] = pd.to_datetime( data['time'], format= '%H:%M:%S')

# time of day
t = pd.to_datetime(data['time'], format = '%H:%M:%S')
secs_since_midnight = t.dt.hour*3600 + t.dt.minute*60 + t.dt.second

# correct for midnight rollover so time keeps increasing
rollovers = (secs_since_midnight.diff() < 0).cumsum()
secs_adjusted = secs_since_midnight + rollovers * 86400

# base date for timestamp to work
base_date = pd.Timestamp('2026-08-16')
data['clock'] = base_date + pd.to_timedelta(secs_adjusted, unit='s')

fig0 = plt.figure()
ax0 = fig0.add_subplot()

# timehrs = (np.linspace(0,8364, 8365) * 5) / 3600
uamps = data['current'] * 1e6

# ax0.plot(data['timestamp'], uamps, '.')
ax0.plot(data['clock'], uamps, '.', markersize = 0.7)

ax0.set_title('Beam Current At-211 Run, August 16th Evening - 17th Morning 2026')
ax0.set_xlabel(r'time (hh:mm:ss)')
ax0.set_ylabel(r'Current ($\mu$A)')
ax0.set_ylim(-0.2,20)
ax0.xaxis.set_major_locator(mdates.HourLocator(interval=1))
ax0.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M:%S'))
fig0.autofmt_xdate()

plt.tight_layout()
```

```python
%matplotlib ipympl 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.dates as mdates

runs = [{'path' : 'C:/users/willm/At-211/Current Data/Current_Aug27morning_2026.dat', 'title' : 'August 16th Evening - 17th Morning 2026', 'base_date' : '2026-08-8'},
        {'path' : 'C:/users/willm/At-211/Current Data/Current_Jul28afternoon_2026.dat', 'title' : 'July 28th 2026', 'base_date' : '2026-07-28'},
        {'path' : 'C:/users/willm/At-211/Current Data/Current_Jul28morning_2026.dat', 'title' : 'July 27th Evening - 28th Morning 2026', 'base_date' : '2026-07-27'},
        {'path' : 'C:/users/willm/At-211/Current Data/CurrentJul_27morning_2026.dat', 'title' : 'July 27th 2026', 'base_date' : '2026-07-27'},
        {'path' : 'C:/users/willm/At-211/Current Data/CurrentJul8_9_2026.dat', 'title' : 'July 8th Evening - 9th Morning 2026', 'base_date' : '2026-07-08'},
        {'path' : 'C:/users/willm/At-211/Current Data/Current_july_8_2026.dat', 'title' : 'July 7th Evening - 8th Morning 2026', 'base_date' : '2026-07-07'},
        {'path' : 'C:/users/willm/At-211/Current Data/Current_Jul6_7_2026.dat', 'title' : 'July 6th Evening - 7th Morning 2026', 'base_date' : '2026-07-06'}]

fig, axs = plt.subplots(4, 2, figsize = (14,16)) # thinnest possible wihtout clumping x-axis labels
axs = axs.flatten()


for i, run in enumerate(runs):

    df = pd.read_csv(run['path'], skiprows = 1, header = None, names = [ 'date', 'time', 'na1','na2','na3', 'current', 'dashes', 'node'], sep = ' ')
    data = df[['date','time','current']].copy()  # add .copy() to prevent warning
    
    # --data['timestamp'] = pd.to_datetime(data['date'] + ' ' + data['time'], format='%m/%d/%Y %H:%M:%S')--
    # --data['timestamp'] = pd.to_datetime( data['time'], format= '%H:%M:%S')--

    # time of day
    t = pd.to_datetime(data['time'], format = '%H:%M:%S')
    secs_since_midnight = t.dt.hour*3600 + t.dt.minute*60 + t.dt.second
    
    # correct for midnight rollover so time keeps increasing
    rollovers = (secs_since_midnight.diff() < 0).cumsum()
    secs_adjusted = secs_since_midnight + rollovers * 86400
    
    # base date and timestamp to make hh:mm:ss labels
    base_date = pd.Timestamp(run['base_date'])
    data['clock'] = base_date + pd.to_timedelta(secs_adjusted, unit='s')
    
    # --timehrs = (np.linspace(0,8364, 8365) * 5) / 3600--
    uamps = data['current'] * 1e6
    
    ax = axs[i]
    
    # --ax0.plot(data['timestamp'], uamps, '.')--
    ax.plot(data['clock'], uamps, '.', markersize = 0.7)
    
    ax.set_title(run['title'])
    ax.set_xlabel(r'time (hh:mm:ss)')

    # turn off ticks for even # plots (right side) 
    if i % 2 == 0:
            ax.set_ylabel(r'Current ($\mu$A)')
    else:
            ax.tick_params(labelleft=False)
    
    ax.set_ylabel(r'Current ($\mu$A)')
    ax.set_ylim(-0.2,20)
    ax.xaxis.set_major_locator(mdates.HourLocator(interval=1))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M:%S'))
    plt.setp(ax.get_xticklabels(), rotation=30, ha='right')   # rotate tick labels per axis
    
axs[7].set_visible(False) # make 8th graph invisible

fig.suptitle('At-211 Runs K150 Beam Current', fontsize = 25)
plt.tight_layout(rect=[0, 0, 1, 0.975]) # leaves 2.5% space after title
```
